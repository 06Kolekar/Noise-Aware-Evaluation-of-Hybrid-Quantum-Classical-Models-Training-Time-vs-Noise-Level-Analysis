```python
# ===============================
# INSTALL (Run once)
# ===============================
!pip install -q qiskit qiskit-aer qiskit-machine-learning qiskit-algorithms

# ===============================
# IMPORTS
# ===============================
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, f1_score

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from qiskit.circuit.library import ZZFeatureMap, RealAmplitudes
from qiskit_algorithms.optimizers import SPSA

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel
from qiskit_aer.noise.errors import depolarizing_error

from qiskit.primitives import BackendSamplerV2 as BackendSampler

from qiskit_machine_learning.algorithms.classifiers import VQC


# ===============================
# MAIN FUNCTION
# ===============================
def run_dataset(file_path, target_col, dataset_name, drop_cols=None, make_binary=False):

    print("\n==============================")
    print("DATASET:", dataset_name)
    print("==============================")

    # Load dataset
    df = pd.read_csv(file_path)

    # Drop unnecessary columns
    if drop_cols:
        df = df.drop(columns=drop_cols)

    # Convert target to binary (for heart dataset)
    if make_binary:
        df[target_col] = df[target_col].apply(lambda x: 1 if x > 0 else 0)

    # Split X and y
    X = df.drop(target_col, axis=1)
    y = df[target_col]
    # Convert categorical labels like B/M into 0/1
    if y.dtype == "object":
      y = y.map({"B": 0, "M": 1})
    y = y.to_numpy()

    # Convert categorical columns in X to numeric
    X = pd.get_dummies(X, drop_first=True)

    # Scaling
    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X)

    # PCA reduce to 2 features (2 qubits)
    pca = PCA(n_components=2)
    X_reduced = pca.fit_transform(X_scaled)

    # Train test split
    X_train, X_test, y_train, y_test = train_test_split(
        X_reduced, y, test_size=0.2, random_state=42
    )

    # ===============================
    # Classical Baselines
    # ===============================
    print("\n--- Classical Models ---")

    start = time.time()
    lr = LogisticRegression()
    lr.fit(X_train, y_train)
    end = time.time()
    pred_lr = lr.predict(X_test)

    print("LR Accuracy:", accuracy_score(y_test, pred_lr))
    print("LR F1:", f1_score(y_test, pred_lr))
    print("LR Time:", end-start)

    start = time.time()
    svm = SVC()
    svm.fit(X_train, y_train)
    end = time.time()
    pred_svm = svm.predict(X_test)

    print("\nSVM Accuracy:", accuracy_score(y_test, pred_svm))
    print("SVM F1:", f1_score(y_test, pred_svm))
    print("SVM Time:", end-start)

    # ===============================
    # Quantum VQC Noise Experiment
    # ===============================
    print("\n--- Quantum VQC (Noise Analysis) ---")

    feature_map = ZZFeatureMap(feature_dimension=2, reps=2).decompose()
    ansatz = RealAmplitudes(num_qubits=2, reps=2).decompose()
    optimizer = SPSA(maxiter=20)

    noise_levels = [0.0, 0.01, 0.03, 0.05, 0.1]
    results = []

    for p in noise_levels:

        # Noise model
        noise_model = NoiseModel()
        error = depolarizing_error(p, 1)
        noise_model.add_all_qubit_quantum_error(error, ['u', 'sx', 'x'])

        # Aer simulator backend with noise
        backend = AerSimulator(noise_model=noise_model)

        # BackendSampler primitive
        sampler = BackendSampler(backend=backend)

        # VQC Model
        vqc = VQC(
            feature_map=feature_map,
            ansatz=ansatz,
            optimizer=optimizer,
            sampler=sampler
        )

        start = time.time()
        vqc.fit(X_train, y_train)
        end = time.time()

        pred_q = vqc.predict(X_test)

        acc = accuracy_score(y_test, pred_q)
        f1 = f1_score(y_test, pred_q)
        train_time = end - start

        results.append([p, acc, f1, train_time])

        print(f"Noise={p}  Accuracy={acc:.4f}  F1={f1:.4f}  Time={train_time:.2f}s")

    df_results = pd.DataFrame(results, columns=["Noise", "Accuracy", "F1", "Training Time"])
    return df_results


# ===============================
# RUN ALL 3 DATASETS
# ===============================

diabetes_results = run_dataset(
    "/content/drive/MyDrive/Colab Notebooks/diabetes.csv",
    "Outcome",
    "Diabetes Dataset"
)

heart_results = run_dataset(
    "/content/drive/MyDrive/Colab Notebooks/cleanned.csv",
    "num",
    "Heart Disease Dataset",
    make_binary=True
)

cancer_results = run_dataset(
    "/content/drive/MyDrive/Colab Notebooks/brca.csv",
    "y",
    "Breast Cancer Dataset",
    drop_cols=["srno"]
)


print("\n=== Diabetes Quantum Results ===")
print(diabetes_results)

print("\n=== Heart Quantum Results ===")
print(heart_results)

print("\n=== Breast Cancer Quantum Results ===")
print(cancer_results)


# ===============================
# PLOT RESULTS
# ===============================
def plot_graphs(df, dataset_name):

    plt.plot(df["Noise"], df["Accuracy"], marker="o")
    plt.xlabel("Noise Level (p)")
    plt.ylabel("Accuracy")
    plt.title(dataset_name + " : Noise vs Accuracy")
    plt.show()

    plt.plot(df["Noise"], df["Training Time"], marker="o")
    plt.xlabel("Noise Level (p)")
    plt.ylabel("Training Time (seconds)")
    plt.title(dataset_name + " : Noise vs Training Time")
    plt.show()


plot_graphs(diabetes_results, "Diabetes Dataset")
plot_graphs(heart_results, "Heart Disease Dataset")
plot_graphs(cancer_results, "Breast Cancer Dataset")
```

```python
!pip uninstall -y qiskit qiskit-aer qiskit-machine-learning qiskit-algorithms
# ================================
# INSTALL PACKAGES (RUN ONCE)
# ================================
!pip install -q qiskit qiskit-aer qiskit-machine-learning qiskit-algorithms


# ================================
# IMPORTS
# ================================
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, f1_score

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from qiskit.circuit.library import ZZFeatureMap, RealAmplitudes
from qiskit_algorithms.optimizers import SPSA

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel
from qiskit_aer.noise.errors import depolarizing_error

from qiskit_machine_learning.algorithms.classifiers import VQC
from qiskit.primitives import BackendSamplerV2

#NEW IMPORT
from concurrent.futures import ProcessPoolExecutor


# ================================
# ZNE FUNCTIONS
# ================================
def zne_linear(v1, v2):
    return (2 * v1) - v2

def zne_richardson(v1, v2, v3):
    return (3 * v1) - (3 * v2) + v3


# ================================
# RUN QUANTUM MODEL FOR ONE NOISE LEVEL
# ================================
def run_vqc_once(X_train, y_train, X_test, y_test, noise_p, optimizer, feature_map, ansatz, shots=1024):

    noise_model = NoiseModel()
    error = depolarizing_error(noise_p, 1)
    noise_model.add_all_qubit_quantum_error(error, ['u', 'sx', 'x'])

    backend = AerSimulator(noise_model=noise_model, shots=shots)
    sampler = BackendSamplerV2(backend)

    vqc = VQC(
        feature_map=feature_map,
        ansatz=ansatz,
        optimizer=optimizer,
        sampler=sampler
    )

    start = time.time()
    vqc.fit(X_train, y_train)
    end = time.time()

    pred = vqc.predict(X_test)

    acc = accuracy_score(y_test, pred)
    f1 = f1_score(y_test, pred)
    train_time = end - start

    return acc, f1, train_time


# ================================
# NEW PARALLEL FUNCTION
# ================================
def process_noise(args):
    p, X_train, y_train, X_test, y_test, optimizer, feature_map, ansatz = args

    acc_base, f1_base, time_base = run_vqc_once(
        X_train, y_train, X_test, y_test,
        noise_p=p,
        optimizer=optimizer,
        feature_map=feature_map,
        ansatz=ansatz
    )

    acc_2p, f1_2p, t_2p = run_vqc_once(
        X_train, y_train, X_test, y_test,
        noise_p=min(2*p, 1.0),
        optimizer=optimizer,
        feature_map=feature_map,
        ansatz=ansatz
    )

    zneA_acc = zne_linear(acc_base, acc_2p)
    zneA_f1 = zne_linear(f1_base, f1_2p)
    zneA_time = (time_base + t_2p) / 2

    acc_3p, f1_3p, t_3p = run_vqc_once(
        X_train, y_train, X_test, y_test,
        noise_p=min(3*p, 1.0),
        optimizer=optimizer,
        feature_map=feature_map,
        ansatz=ansatz
    )

    zneB_acc = zne_richardson(acc_base, acc_2p, acc_3p)
    zneB_f1 = zne_richardson(f1_base, f1_2p, f1_3p)
    zneB_time = (time_base + t_2p + t_3p) / 3

    print(f"Noise={p} | BaseAcc={acc_base:.4f} | ZNE-A={zneA_acc:.4f} | ZNE-B={zneB_acc:.4f}")

    return (
        [p, acc_base, f1_base, time_base],
        [p, zneA_acc, zneA_f1, zneA_time],
        [p, zneB_acc, zneB_f1, zneB_time]
    )


# ================================
# MAIN DATASET FUNCTION
# ================================
def run_dataset_full(file_path, target_col, dataset_name, drop_cols=None, make_binary=False, pca_components=2):

    print("\n==============================")
    print("DATASET:", dataset_name)
    print("==============================")

    df = pd.read_csv(file_path)

    if drop_cols:
        df = df.drop(columns=drop_cols)

    y = df[target_col]

    if y.dtype == "object":
        y = y.map({"B": 0, "M": 1})

    if make_binary:
        y = y.apply(lambda x: 1 if int(x) > 0 else 0)

    y = y.to_numpy()

    X = df.drop(target_col, axis=1)
    X = pd.get_dummies(X, drop_first=True)

    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X)

    pca = PCA(n_components=pca_components)
    X_reduced = pca.fit_transform(X_scaled)

    X_train, X_test, y_train, y_test = train_test_split(
        X_reduced, y, test_size=0.2, random_state=42
    )

    print("\n--- Classical Models ---")

    start = time.time()
    lr = LogisticRegression()
    lr.fit(X_train, y_train)
    end = time.time()
    pred_lr = lr.predict(X_test)

    lr_acc = accuracy_score(y_test, pred_lr)
    lr_f1 = f1_score(y_test, pred_lr)
    lr_time = end - start

    print("LR Accuracy:", lr_acc)

    start = time.time()
    svm = SVC()
    svm.fit(X_train, y_train)
    end = time.time()
    pred_svm = svm.predict(X_test)

    svm_acc = accuracy_score(y_test, pred_svm)

    print("\n--- Quantum VQC (Parallel + ZNE) ---")

    feature_map = ZZFeatureMap(feature_dimension=pca_components, reps=2).decompose()
    ansatz = RealAmplitudes(num_qubits=pca_components, reps=2).decompose()

    optimizer = SPSA(maxiter=50)

    noise_levels = [0.0, 0.01, 0.03, 0.05, 0.1]

    # ================================
    # PARALLEL EXECUTION
    # ================================
    args_list = [
        (p, X_train, y_train, X_test, y_test, optimizer, feature_map, ansatz)
        for p in noise_levels
    ]

    baseline_results = []
    zneA_results = []
    zneB_results = []

    with ProcessPoolExecutor(max_workers=2) as executor:
        results = list(executor.map(process_noise, args_list))

    for base, zA, zB in results:
        baseline_results.append(base)
        zneA_results.append(zA)
        zneB_results.append(zB)

    df_base = pd.DataFrame(baseline_results, columns=["Noise", "Accuracy", "F1", "Time"])
    df_zneA = pd.DataFrame(zneA_results, columns=["Noise", "Accuracy", "F1", "Time"])
    df_zneB = pd.DataFrame(zneB_results, columns=["Noise", "Accuracy", "F1", "Time"])

    return df_base, df_zneA, df_zneB


# ================================
# RUN ALL DATASETS
# ================================
if __name__ == "__main__":

    diabetes_base, diabetes_zneA, diabetes_zneB = run_dataset_full(
        "/content/drive/MyDrive/Colab Notebooks/diabetes.csv",
        "Outcome",
        "Diabetes Dataset"
    )

    heart_base, heart_zneA, heart_zneB = run_dataset_full(
        "/content/drive/MyDrive/Colab Notebooks/cleanned.csv",
        "num",
        "Heart Disease Dataset",
        make_binary=True
    )

    cancer_base, cancer_zneA, cancer_zneB = run_dataset_full(
        "/content/drive/MyDrive/Colab Notebooks/brca.csv",
        "y",
        "Breast Cancer Dataset",
        drop_cols=["srno"]
    )

    print("\nDone.")
```
