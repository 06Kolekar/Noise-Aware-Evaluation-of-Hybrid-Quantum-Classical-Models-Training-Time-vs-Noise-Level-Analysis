```python
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, f1_score

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from qiskit.circuit.library import ZZFeatureMap, RealAmplitudes
from qiskit_algorithms.optimizers import SPSA

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel
from qiskit_aer.noise.errors import depolarizing_error

from qiskit.primitives import BackendSamplerV2
from qiskit_machine_learning.algorithms.classifiers import VQC

from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit import transpile

print("Qiskit version check done!")


# ============================
# ZNE FUNCTIONS
# ============================
def zne_linear(v1, v2):
    # Extrapolate to noise=0 using two points
    return (2 * v1) - v2

def zne_richardson(v1, v2, v3):
    # Richardson extrapolation (3 points)
    return (3 * v1) - (3 * v2) + v3


# ============================
# TRANSPILER OPTIMIZATION
# ============================
def optimize_circuit(circuit, backend, optimization_level=3):
    """
    Uses Qiskit preset pass manager to optimize the circuit
    (layout + routing + translation + optimization).
    """
    pm = generate_preset_pass_manager(
        optimization_level=optimization_level,
        backend=backend
    )
    optimized_circuit = pm.run(circuit)
    return optimized_circuit


# ============================
# RUN VQC ONCE
# ============================
def run_vqc_once(X_train, y_train, X_test, y_test,
                noise_p,
                optimizer,
                feature_map,
                ansatz,
                shots=1024,
                optimization_level=3):

    # ----------------------------
    # Noise Model
    # ----------------------------
    noise_model = NoiseModel()
    error = depolarizing_error(noise_p, 1)
    noise_model.add_all_qubit_quantum_error(error, ['u', 'sx', 'x'])

    backend = AerSimulator(noise_model=noise_model)

    # ----------------------------
    # Create full circuit (feature_map + ansatz)
    # ----------------------------
    qc = feature_map.compose(ansatz)
    qc.measure_all()

    # ----------------------------
    # Transpile Optimization
    # ----------------------------
    qc_opt = optimize_circuit(qc, backend, optimization_level=optimization_level)

    # ----------------------------
    # Sampler
    # ----------------------------
    sampler = BackendSamplerV2(backend=backend)
    sampler.options.shots = shots

    # ----------------------------
    # VQC
    # ----------------------------
    vqc = VQC(
        feature_map=feature_map,
        ansatz=ansatz,
        optimizer=optimizer,
        sampler=sampler
    )

    # ----------------------------
    # Train
    # ----------------------------
    start = time.time()
    vqc.fit(X_train, y_train)
    end = time.time()

    pred = vqc.predict(X_test)

    acc = accuracy_score(y_test, pred)
    f1 = f1_score(y_test, pred)

    return acc, f1, (end - start)


# ============================
# RUN FULL DATASET
# ============================
def run_dataset_full(file_path, target_col, dataset_name,
                     drop_cols=None,
                     make_binary=False,
                     pca_components=2,
                     maxiter=20,
                     shots=1024,
                     optimization_level=3):

    print("\n==============================")
    print("DATASET:", dataset_name)
    print("==============================")

    df = pd.read_csv(file_path)

    if drop_cols:
        df = df.drop(columns=drop_cols)

    y = df[target_col]

    # Breast Cancer conversion
    if y.dtype == "object":
        y = y.map({"B": 0, "M": 1})

    # Heart Disease conversion
    if make_binary:
        y = y.apply(lambda x: 1 if int(x) > 0 else 0)

    y = y.to_numpy()

    X = df.drop(target_col, axis=1)

    # Convert categorical columns (Male/Female)
    X = pd.get_dummies(X, drop_first=True)

    # Scaling
    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X)

    # PCA
    pca = PCA(n_components=pca_components)
    X_reduced = pca.fit_transform(X_scaled)

    # Split
    X_train, X_test, y_train, y_test = train_test_split(
        X_reduced, y, test_size=0.2, random_state=42
    )

    # ----------------------------
    # CLASSICAL MODELS
    # ----------------------------
    print("\n--- Classical Models ---")

    lr = LogisticRegression()
    lr.fit(X_train, y_train)
    pred_lr = lr.predict(X_test)
    lr_acc = accuracy_score(y_test, pred_lr)

    svm = SVC()
    svm.fit(X_train, y_train)
    pred_svm = svm.predict(X_test)
    svm_acc = accuracy_score(y_test, pred_svm)

    print("LR Accuracy:", lr_acc)
    print("SVM Accuracy:", svm_acc)

    # ----------------------------
    # QUANTUM SETTINGS
    # ----------------------------
    print("\n--- Quantum VQC (Baseline + ZNE + Optimized Transpiler) ---")

    feature_map = ZZFeatureMap(feature_dimension=pca_components, reps=2).decompose()
    ansatz = RealAmplitudes(num_qubits=pca_components, reps=2).decompose()

    optimizer = SPSA(maxiter=maxiter)

    noise_levels = [0.0, 0.01, 0.03, 0.05, 0.1]

    baseline_results = []
    zneA_results = []
    zneB_results = []

    for p in noise_levels:

        # BASELINE (p)
        acc_base, f1_base, t_base = run_vqc_once(
            X_train, y_train, X_test, y_test,
            noise_p=p,
            optimizer=optimizer,
            feature_map=feature_map,
            ansatz=ansatz,
            shots=shots,
            optimization_level=optimization_level
        )
        baseline_results.append([p, acc_base, f1_base, t_base])

        # Run at 2p
        acc_2p, f1_2p, t_2p = run_vqc_once(
            X_train, y_train, X_test, y_test,
            noise_p=min(2*p, 1.0),
            optimizer=optimizer,
            feature_map=feature_map,
            ansatz=ansatz,
            shots=shots,
            optimization_level=optimization_level
        )

        # ZNE-A
        zneA_acc = zne_linear(acc_base, acc_2p)
        zneA_f1 = zne_linear(f1_base, f1_2p)
        zneA_time = (t_base + t_2p) / 2
        zneA_results.append([p, zneA_acc, zneA_f1, zneA_time])

        # Run at 3p
        acc_3p, f1_3p, t_3p = run_vqc_once(
            X_train, y_train, X_test, y_test,
            noise_p=min(3*p, 1.0),
            optimizer=optimizer,
            feature_map=feature_map,
            ansatz=ansatz,
            shots=shots,
            optimization_level=optimization_level
        )

        # ZNE-B
        zneB_acc = zne_richardson(acc_base, acc_2p, acc_3p)
        zneB_f1 = zne_richardson(f1_base, f1_2p, f1_3p)
        zneB_time = (t_base + t_2p + t_3p) / 3
        zneB_results.append([p, zneB_acc, zneB_f1, zneB_time])

        print(f"Noise={p} | BaseAcc={acc_base:.4f} | ZNE-A={zneA_acc:.4f} | ZNE-B={zneB_acc:.4f}")

    df_base = pd.DataFrame(baseline_results, columns=["Noise", "Accuracy", "F1", "Training Time"])
    df_zneA = pd.DataFrame(zneA_results, columns=["Noise", "Accuracy", "F1", "Training Time"])
    df_zneB = pd.DataFrame(zneB_results, columns=["Noise", "Accuracy", "F1", "Training Time"])

    # ============================
    # GRAPH 1: Noise vs Accuracy
    # ============================
    plt.figure(figsize=(10,5))
    plt.plot(df_base["Noise"], df_base["Accuracy"], marker="o", label="Baseline VQC")
    plt.plot(df_zneA["Noise"], df_zneA["Accuracy"], marker="o", label="ZNE-A Linear")
    plt.plot(df_zneB["Noise"], df_zneB["Accuracy"], marker="o", label="ZNE-B Richardson")
    plt.axhline(y=lr_acc, linestyle="--", label="Classical LR")
    plt.axhline(y=svm_acc, linestyle="--", label="Classical SVM")
    plt.xlabel("Noise Level (p)")
    plt.ylabel("Accuracy")
    plt.title(dataset_name + " : Noise vs Accuracy")
    plt.legend()
    plt.show()

    # ============================
    # GRAPH 2: Noise vs Training Time
    # ============================
    plt.figure(figsize=(10,5))
    plt.plot(df_base["Noise"], df_base["Training Time"], marker="o", label="Baseline VQC")
    plt.plot(df_zneA["Noise"], df_zneA["Training Time"], marker="o", label="ZNE-A Time")
    plt.plot(df_zneB["Noise"], df_zneB["Training Time"], marker="o", label="ZNE-B Time")
    plt.xlabel("Noise Level (p)")
    plt.ylabel("Training Time (seconds)")
    plt.title(dataset_name + " : Noise vs Training Time")
    plt.legend()
    plt.show()

    return df_base, df_zneA, df_zneB, lr_acc, svm_acc


# ================================
# RUN ALL DATASETS
# ================================
diabetes_base, diabetes_zneA, diabetes_zneB, diabetes_lr, diabetes_svm = run_dataset_full(
    "/content/drive/MyDrive/Colab Notebooks/diabetes.csv",
    "Outcome",
    "Diabetes Dataset",
    pca_components=2,
    maxiter=20,
    shots=1024,
    optimization_level=3
)

heart_base, heart_zneA, heart_zneB, heart_lr, heart_svm = run_dataset_full(
    "/content/drive/MyDrive/Colab Notebooks/cleanned.csv",
    "num",
    "Heart Disease Dataset",
    make_binary=True,
    pca_components=2,
    maxiter=20,
    shots=1024,
    optimization_level=3
)

cancer_base, cancer_zneA, cancer_zneB, cancer_lr, cancer_svm = run_dataset_full(
    "/content/drive/MyDrive/Colab Notebooks/brca.csv",
    "y",
    "Breast Cancer Dataset",
    drop_cols=["srno"],
    pca_components=2,
    maxiter=20,
    shots=1024,
    optimization_level=3
)


# ================================
# COMBINED TABLES
# ================================
print("\n==============================")
print("COMBINED RESULTS TABLES")
print("==============================")

combined_accuracy = pd.DataFrame({
    "Noise": diabetes_base["Noise"],
    "Diabetes_Base": diabetes_base["Accuracy"],
    "Diabetes_ZNE-A": diabetes_zneA["Accuracy"],
    "Diabetes_ZNE-B": diabetes_zneB["Accuracy"],

    "Heart_Base": heart_base["Accuracy"],
    "Heart_ZNE-A": heart_zneA["Accuracy"],
    "Heart_ZNE-B": heart_zneB["Accuracy"],

    "Cancer_Base": cancer_base["Accuracy"],
    "Cancer_ZNE-A": cancer_zneA["Accuracy"],
    "Cancer_ZNE-B": cancer_zneB["Accuracy"]
})

combined_time = pd.DataFrame({
    "Noise": diabetes_base["Noise"],
    "Diabetes_Base_Time": diabetes_base["Training Time"],
    "Diabetes_ZNE-A_Time": diabetes_zneA["Training Time"],
    "Diabetes_ZNE-B_Time": diabetes_zneB["Training Time"],

    "Heart_Base_Time": heart_base["Training Time"],
    "Heart_ZNE-A_Time": heart_zneA["Training Time"],
    "Heart_ZNE-B_Time": heart_zneB["Training Time"],

    "Cancer_Base_Time": cancer_base["Training Time"],
    "Cancer_ZNE-A_Time": cancer_zneA["Training Time"],
    "Cancer_ZNE-B_Time": cancer_zneB["Training Time"]
})

print("\n=== Combined Accuracy Table ===")
print(combined_accuracy)

print("\n=== Combined Training Time Table ===")
print(combined_time)


# ================================
# FINAL COMPARISON GRAPHS (ALL DATASETS TOGETHER)
# ================================
plt.figure(figsize=(12,6))
plt.plot(combined_accuracy["Noise"], combined_accuracy["Diabetes_Base"], marker="o", label="Diabetes Base")
plt.plot(combined_accuracy["Noise"], combined_accuracy["Heart_Base"], marker="o", label="Heart Base")
plt.plot(combined_accuracy["Noise"], combined_accuracy["Cancer_Base"], marker="o", label="Cancer Base")
plt.xlabel("Noise Level (p)")
plt.ylabel("Accuracy")
plt.title("Baseline VQC Accuracy Comparison (All Datasets)")
plt.legend()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(combined_time["Noise"], combined_time["Diabetes_Base_Time"], marker="o", label="Diabetes Base Time")
plt.plot(combined_time["Noise"], combined_time["Heart_Base_Time"], marker="o", label="Heart Base Time")
plt.plot(combined_time["Noise"], combined_time["Cancer_Base_Time"], marker="o", label="Cancer Base Time")
plt.xlabel("Noise Level (p)")
plt.ylabel("Training Time (seconds)")
plt.title("Baseline VQC Training Time Comparison (All Datasets)")
plt.legend()
plt.show()


print("\n=== ALL DONE SUCCESSFULLY ===")
```
